```python
import numpy as np
from scipy.optimize import brentq # Para encontrar la raíz de la ecuación de restricción

def calcular_eoq_sin_restriccion(D, Co, Cc):
    """Calcula la EOQ no restringida para un solo artículo."""
    # EOQ_i* = sqrt((2 * Co_i * D_i) / Cc_i)
    return np.sqrt((2 * Co * D) / Cc)

def calcular_cantidad_con_restriccion(D, Co, Cc, a, lambda_val):
    """Calcula la cantidad de pedido (Y_i) para un artículo con el multiplicador lambda."""
    # Y_i = sqrt((2 * Co_i * D_i) / (Cc_i + 2 * lambda * a_i))
    # Se debe asegurar que el denominador no sea negativo. Como Cc_i > 0 y lambda >= 0 y a_i > 0, esto siempre se cumple.
    return np.sqrt((2 * Co * D) / (Cc + 2 * lambda_val * a))

def funcion_restriccion_lambda(lambda_val, D, Co, Cc, a, A):
    """
    Define la función para encontrar el lambda. Buscamos la raíz (donde la función es 0).
    f(lambda) = Sumatoria(a_i * Y_i(lambda)) - A = 0
    """
    if lambda_val < 0:
        # El multiplicador de Lagrange debe ser no negativo
        return np.inf

    Y_i = calcular_cantidad_con_restriccion(D, Co, Cc, a, lambda_val)
    espacio_total_usado = np.sum(a * Y_i)

    # Queremos encontrar lambda tal que el espacio usado sea igual a la capacidad (A)
    return espacio_total_usado - A

def eoq_multi_item_almacen(D, Co, Cc, a, A):
    """
    Implementa el modelo EOQ de múltiples artículos con restricción de almacén.

    Args:
        D (array): Demandas anuales de los artículos.
        Co (array): Costos de ordenar por pedido.
        Cc (array): Costos de conservación por unidad.
        a (array): Espacio de almacén por unidad.
        A (float): Capacidad total de almacén.

    Returns:
        tuple: (Q_opt, lambda_opt, eoq_sin_restriccion)
    """

    # 1. Calcular las EOQ sin restricción
    Y_star = calcular_eoq_sin_restriccion(D, Co, Cc)
    espacio_sin_restriccion = np.sum(a * Y_star)

    print(f"Capacidad requerida sin restricción: {espacio_sin_restriccion:.2f}")
    print(f"Capacidad del almacén (A): {A:.2f}")

    # 2. Verificar la restricción
    if espacio_sin_restriccion <= A:
        # No hay restricción activa, la solución óptima es la EOQ no restringida
        print("La restricción no está activa. La solución es la EOQ sin restricción.")
        return Y_star, 0, Y_star
    else:
        print("La restricción está activa. Se calcula el multiplicador de Lagrange (lambda).")

        # 3. Encontrar el valor de lambda (usando brentq, que busca la raíz)
        # Necesitamos un rango [a, b] donde f(a) y f(b) tengan signos opuestos.
        # f(0) debe ser positivo (espacio_sin_restriccion - A > 0)
        # Para un lambda muy grande, Y_i -> 0, por lo que f(lambda) será negativo (0 - A < 0)

        lambda_inicial_b = 1.0 # Empezamos la búsqueda con un valor pequeño
        while funcion_restriccion_lambda(lambda_inicial_b, D, Co, Cc, a, A) > 0:
            lambda_inicial_b *= 2
            if lambda_inicial_b > 1e10:
                 # Límite de seguridad
                 raise Exception("No se pudo encontrar un límite superior para lambda.")

        # brentq encuentra la raíz en el intervalo [0, lambda_inicial_b]
        lambda_opt = brentq(funcion_restriccion_lambda, 0, lambda_inicial_b, args=(D, Co, Cc, a, A))

        # 4. Calcular las cantidades óptimas restringidas
        Y_opt = calcular_cantidad_con_restriccion(D, Co, Cc, a, lambda_opt)

        print(f"Multiplicador de Lagrange óptimo (lambda): {lambda_opt:.4f}")
        print(f"Espacio total usado con Y_opt: {np.sum(a * Y_opt):.2f}")
        return Y_opt, lambda_opt, Y_star

# --- EJEMPLO DE USO (Adaptado de la metodología común del libro de Taha) ---

# Parámetros de ejemplo para 3 artículos
#               Artículo 1  Artículo 2  Artículo 3
D_ejemplo = np.array([1000, 2000, 3000])  # Demanda anual (unidades/año)
Co_ejemplo = np.array([100, 150, 80])    # Costo de ordenar/preparar ($/pedido)
Cc_ejemplo = np.array([0.5, 0.4, 0.6])   # Costo de conservación ($/unidad/año)
a_ejemplo = np.array([2, 3, 1])          # Espacio requerido (pies cúbicos/unidad)

A_ejemplo = 5000.0                       # Capacidad máxima del almacén (pies cúbicos)

# Ejecutar el modelo
Q_opt, lambda_opt, Q_sin_restriccion = eoq_multi_item_almacen(
    D_ejemplo, Co_ejemplo, Cc_ejemplo, a_ejemplo, A_ejemplo
)

# Mostrar resultados
print("\n### Resultados ###")
print(f"| Artículo | EOQ sin Restricción (Y*i) | Cantidad Óptima Restringida (Yi) |")
print("|----------|---------------------------|----------------------------------|")
for i in range(len(D_ejemplo)):
    print(f"| {i+1:8} | {Q_sin_restriccion[i]:25.2f} | {Q_opt[i]:30.2f} |")

# El concepto de la penalización por restricción se puede visualizar
```
